# 03 — Baseline Models
 (ML Engineer)

Trains simple baseline classifiers for `job_ready` (binary) to establish
a performance floor before trying more complex models in
`04_model_comparison.ipynb`.

```python
import sys
sys.path.append("../src")

from preprocessing import run_preprocessing
from evaluate import evaluate_binary_classifier
from sklearn.linear_model import LogisticRegression
from sklearn.dummy import DummyClassifier

(X_train, X_test, y_job_train, y_job_test,
 y_track_train, y_track_test, preprocessor) = run_preprocessing()

X_train_t = preprocessor.transform(X_train)
X_test_t = preprocessor.transform(X_test)
```

## Dummy baseline (predicts the majority class)

Any real model MUST beat this — otherwise it has learned nothing.

```python
dummy = DummyClassifier(strategy="most_frequent", random_state=42)
dummy.fit(X_train_t, y_job_train)
dummy_metrics = evaluate_binary_classifier(dummy, X_test_t, y_job_test)
{k: v for k, v in dummy_metrics.items() if k != "confusion_matrix"}
```

## Logistic Regression baseline (interpretable)

```python
logreg = LogisticRegression(max_iter=1000, random_state=42)
logreg.fit(X_train_t, y_job_train)
logreg_metrics = evaluate_binary_classifier(logreg, X_test_t, y_job_test)
{k: v for k, v in logreg_metrics.items() if k != "confusion_matrix"}
```

**Observation:** Logistic Regression should comfortably beat the dummy
baseline on ROC-AUC and F1 — confirming there is real, learnable signal
in the features (not just class imbalance).
